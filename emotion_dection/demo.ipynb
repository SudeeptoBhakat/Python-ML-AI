```python
import os
import numpy as np
import matplotlib.pyplot as plt
import pickle
import cv2
```

```python
img_dir = os.path.join(os.getcwd(), "Emotion")
img_dir 
```

```python
img_final = os.path.join(os.getcwd(), 'data')
```

```python
images = [] # to store the images
labels = [] # to store the lables
```

```python
for i in os.listdir(img_dir): 
    for j in os.listdir(os.path.join(img_dir, i)):
        img_path = os.path.join(os.path.join(img_dir, i), j) #path of the image 
        img = cv2.imread(img_path) # reading the images 
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # converting the images into grayscale
        img = cv2.resize(img, (48, 48)) # resizing images to 48x48

        images.append(img) # appending the images to a list
        # labels.append(i) # appending the lables to the list 
        labels.append(i.split('.')[0]) # 
    print(i, "--> Completed")
```

```python
set(labels)
```

```python
images = np.array(images)
labels = np.array(labels)
print(len(images), len(labels))
```

```python
plt.imshow(images[6200], cmap='gray')
plt.show()
```

```python
with open(os.path.join(img_final, 'images.p'), 'wb') as f:
    pickle.dump(images, f)

with open(os.path.join(img_final, 'labels.p'), 'wb') as f:
    pickle.dump(labels, f)
```

```python
with open('data\images.p', 'rb') as f:
    x = pickle.load(f)

with open('data\labels.p', 'rb') as f:
    y = pickle.load(f)
```

```python
print("Images: ", x.shape)
print("Labels: ", y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

y = le.fit_transform(y)
```

```python
y
```

```python
x = x / 255
x = x.reshape(-1, 48, 48, 1)
```

```python
print(y.shape)
```

```python
from keras.utils import to_categorical
from tensorflow.keras.utils import to_categorical
```

```python
y = to_categorical(y)
```

```python
y
```

```python
print(len(x))
print(len(y))

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.1)
```

```python
print("x_train", x_train.shape)
print("x_test", x_test.shape)
print("y_train", y_train.shape)
print("y_train", y_test.shape)
```

```python
from keras.models import Sequential
from keras.optimizers import Adam
from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten
```

```python
model = Sequential()

# convoluational layer
model.add(Conv2D(45, (3, 3), activation = 'relu', input_shape = (48, 48, 1)))
model.add(MaxPooling2D(pool_size = (2, 2)))

# 2nd convoluational layer
model.add(Conv2D(64, (3, 3), activation = 'relu'))
model.add(MaxPooling2D(pool_size = (2, 2)))

# 3nd convoluational layer
model.add(Conv2D(128, (3, 3), activation = 'relu'))
model.add(MaxPooling2D(pool_size = (2, 2)))

# Adding flatten layer - converts 2d to 1d
model.add(Flatten())

# Hidden Layer
model.add(Dense(256, activation='relu'))

# Output layer
model.add(Dense(5, activation='softmax'))
model.compile(Adam(learning_rate=0.001), loss = "categorical_crossentropy", metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(x_train, y_train, epochs = 20, validation_data=(x_test, y_test))
```

```python
model.save('emotion_detection_model.h5')
print("Model Saved")
```
