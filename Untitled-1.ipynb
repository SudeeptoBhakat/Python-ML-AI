```python
import re
def clean_text(text):
    text = text.lower().strip()

    # normalize placeholders: [Money] → <money>
    text = re.sub(r"\[(.*?)\]", r"<\1>", text)

    # remove special chars but keep $, < >
    text = re.sub(r"[^a-z0-9<>$ ]", "", text)

    # normalize spaces
    text = re.sub(r"\s+", " ", text)

    return text

df["clean_text"] = df["text"].apply(clean_text)
df.drop('text', axis=1)
```
