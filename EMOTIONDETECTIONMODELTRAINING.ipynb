```python
#IMPORT PACKAGES
import pickle
import keras
import tensorflow
import pandas as pd
import numpy as np
```

```python
#loading the data
#rb - read binary
#wb - write binary
with open("/content/images.p","rb") as f:
  x = pickle.load(f)
with open("/content/labels.p","rb") as f:
  y = pickle.load(f)
```

```python
print("Images:",x.shape)
print("Labels:",y.shape)
```

```python
#Encoding string data in labels:
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
y
```

```python
#standardizing image data:
x = x/255
x = x.reshape(-1,48,48,1)
```

```python
# -1 = total number of data in x variable, 48,48 = size of image, 1 = channel(grayscale)
```

```python
#using one hot encoding to output column
from keras.utils import to_categorical
y = to_categorical(y)
```

```python
#Spliting of training and testing data:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1)
```

```python
print("x_train",x_train.shape)
print("x_test",x_test.shape)
print("y_train",y_train.shape)
print("y_test",y_test.shape)
```

```python
#CNN model
from keras.models import Sequential
from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten
from keras.optimizers import Adam
```

```python
# Con2v  = create a convolutional layer(extract import features), MaxPooling2d = compress the array size, Flatten = it will convert 2d metrix to 1d metrics
```

```python
model = Sequential()
model.add(Conv2D(28,(3,3),activation='relu',input_shape=(48,48,1)))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(56,(3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(112,(3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))

#Flatten layer
model.add(Flatten())

#hidden Layer:
model.add(Dense(256,activation='relu'))

#output Layer
model.add(Dense(5,activation='softmax'))
model.compile(Adam(learning_rate=0.001),loss="categorical_crossentropy",metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(x_train,y_train,epochs=30,validation_data=(x_test,y_test))
```

```python
#to save the model after training
#".h5" = deep learning models are save using .h5 extension
#.h5 = HDF5 file format
#It helps to store/save: model architecture ,weights, optimizers
#training configuaration
model.save("emotion_cnn_model.h5")
print("model saved")
```
