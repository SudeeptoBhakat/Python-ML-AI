```python
#face detection
img = cv2.imread(r"images.jpg")

plt.imshow(img[::,::,::-1])
plt.grid()
plt.show()
```

```python
#face detection using haarcascade
face = r"haarcascade_frontalface_default.xml" #haarcascade copy path
model = cv2.CascadeClassifier(face)

img= cv2.imread(r"face.jpg") #image copy path
facepoints = model.detectMultiScale(img)
for x,y,w,h in facepoints:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
#face detection using web cam
face = r"haarcascade_frontalface_default.xml"
cap = cv2.VideoCapture(0)
ret, frame = cap.read()

while ret:
    ret, frame = cap.read()
    if not ret:
        break
    model = cv2.CascadeClassifier(face)
    facapoints = model.detectMultiScale(frame)
    for x,y,w,h in facapoints:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(0,0,255),3)
    cv2.imshow("Face",frame)
    if cv2.waitKey(10)==ord('q'):
        break
cv2.destroyAllWindows()
cap.release()
        
```
