## vectorizer

<pre>Vertorizer is the process of converting words into numbers . It is a methology in NLP to map words from vocabulary to a corresponding vector of a real numbers which is used to find word predictions , similarities etc.

it is two types:
    --> Count vectorizer
    --> Tf-Idf vectorizer
</pre>

## count vectorizer
count vectorizer means breaking down a sentence or any task into words by performing tasks like converting all words to lowercase, thus removing special characters.

in NLP models cannot understand textual data they only accept numbers so this textual data needs to be vectorized. 

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
txt =["i love sleeping and i love painting"]
x = cv.fit_transform(txt)
```

```python
print(x)
```

```python
print(cv.vocabulary_)
```

```python
print(x.toarray)
```

```python
txt =["i love sleeping and i love painting","i love Biriyani"]
y = cv.fit_transform(txt)
```

```python
y
```

```python
print(cv.vocabulary_)
```

```python
print(y.toarray())
```

## TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
txt = ['The ant killed the elephant']
z = tfidf.fit_transform(txt)
```

```python
z
```

```python
print(z)
```

```python
print(tfidf.vocabulary_)
```

```python
print(z.toarray())
```
